## Herramientas

| Herramienta | Rol en este proyecto |
|---|---|
| **SQLite** (`sqlite3`) | File-based SQL database engine, no server needed |
| **pandas** | Reads the CSV data and writes it to SQLite with `DataFrame.to_sql()` |
| **JupySQL** (`%sql`) | Runs SQL queries directly in notebook cells |

## Fuentes de datos

```python
import prettytable 
import pandas as pd
import sqlite3
```

## Conexión y carga a la base de datos

```python
conn = sqlite3.connect("chicago.db")
prettytable.DEFAULT = 'DEFAULT'
```

```python
%load_ext sql
```

1. Almacenar los datos del `.CSV` en la variable `df1` que hace referencia a DataFrame1.
2. Convertir el Dataframe a SQL, nombrar la tabla y tomar de referencia la base de datos de sqlite3 `chicago.db` almacenada en la variable `conn`.
3. Repetir los pasos anteriores para los DataSets posteriores.

```python
df1 = pd.read_csv("https://data.cityofchicago.org/resource/9xs2-f89t.csv")
df1.to_sql("chicago_schools", conn, if_exists='replace', index=False)
```

```python
df2 = pd.read_csv("https://data.cityofchicago.org/resource/kn9c-c2s2.csv") 
df2.to_sql("chicago_census", conn, if_exists='replace', index=False)
```

Para el DATASET "Crimes" se ajusto un limite a 1000 registros (`?$limit=1000`) para no acceder a todos los registros ya que actualmente son 8,653,072 registros, puedes probar con los que consideres necesarios.

```python
df3 = pd.read_csv("https://data.cityofchicago.org/resource/ijzp-q8t2.csv?$limit=1000")
df3.to_sql("chicago_crimes", conn, if_exists='replace', index=False)
```

Conexión de MAGIC SQL a la base de datos que tenemos en `SQLITE3`.

```python
%sql sqlite:///chicago.db
```

## BUSQUEDAS Y ANALISIS.
 Ejemplo de una busqueda: 
` %sql SELECT * FROM Chicago_census `


1. Encontrar nombre de la columna y tipo de dato de `Chicago_crimes`

```python
%sql PRAGMA table_info(chicago_crimes);
```

2. Encontrar el numero de crimenes de la tabla `chicago_crimes`

```python
%sql SELECT COUNT(*) AS total_crimes_registered FROM chicago_crimes
```

3. Encontrar crimenes menores 

```python
%sql SELECT * FROM chicago_crimes WHERE `description` like '%minor%' limit 5
```

4. Encontrar crimenes que sucedieron en escuelas

```python
%sql SELECT * From Chicago_crimes WHERE `LOCATION_DESCRIPTION` LIKE '%SCHOOL%'
```

5. Encontrar nombre de la columna y tipo de dato de `Chicago_schools`

```python
%sql PRAGMA table_info(Chicago_schools);
```

6. Encontrar nombre de la columna y tipo de dato de `Chicago_census`

```python
%sql PRAGMA table_info(chicago_census)
```

7. Enumera cinco áreas comunitarias con el mayor % de hogares por debajo de la línea de pobreza.

```python
%sql SELECT `Community_area_name`, `PERCENT_HOUSEHOLDS_BELOW_POVERTY` FROM Chicago_census ORDER BY `PERCENT_HOUSEHOLDS_BELOW_POVERTY` DESC LIMIT 5
```
